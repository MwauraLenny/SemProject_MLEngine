# 1. Dataset Discovery
**Research Area:** Financial Analysis of Collective Investment Schemes (CIS) in Kenya.

**Dataset Identified:** Kenyan Collective Investment Schemes Dataset (via Hugging Face).

**Relevance:** This dataset contains historical records of various investment schemes, including entries for fund values and dates. It is relevant for building an ML application that tracks or predicts investment growth and fund performance for users.

# 2. Expected Output Variable
For the final application, the **Expected Output (Target Variable)** will be the `entry_value` (representing the fund's monetary value or performance metric).

**Rationale:** The ML engine will aim to predict the future value of an investment scheme based on historical entries. By targeting `entry_value`, the mobile app can provide users with insights into potential returns for their chosen investment scheme.
